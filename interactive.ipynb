```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt

n_rows, n_cols = 25, 25          # board size
cells = np.zeros((n_rows, n_cols))

fig, ax = plt.subplots()
image = ax.imshow(cells, cmap="binary", vmin=0, vmax=1)
ax.set_xticks([]); ax.set_yticks([])

def on_click(event):
    if event.inaxes != ax:
        return
    col = int(round(event.xdata))   # x on the image is the column
    row = int(round(event.ydata))   # y on the image is the row
    cells[row, col] = 1 - cells[row, col]
    image.set_data(cells)
    fig.canvas.draw_idle()

fig.canvas.mpl_connect("button_press_event", on_click)
plt.show()
```

```python
from jcgol import GameOfLife
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

steps = 20

live = [(int(r), int(c)) for r, c in np.argwhere(cells == 1)]
gol = GameOfLife(n_rows, n_cols, live)
gol.run(steps)

fig2, ax2 = plt.subplots()
image2 = ax2.imshow(gol.history_[0], cmap="binary", vmin=0, vmax=1)
ax2.set_xticks([]); ax2.set_yticks([])

def update(frame):
    image2.set_data(gol.history_[frame])
    return [image2]

animation = FuncAnimation(fig2, update, frames=len(gol.history_), interval=100, blit=True)
plt.close(fig2)   # keeps an extra blank figure from appearing
HTML(animation.to_jshtml())
```
