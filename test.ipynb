```python
import numpy as np
```

```python
board = np.zeros((3,4), dtype = int)

for row in range(3):
    for col in range (4):
        rand = np.random.randint(10)
        if rand > 5:
            board[row][col] = 1
board[0][0] = 2

print(f"board[0][0] \n  {board[0][0]} ")

print(f"board:  \n {board}")

p_board = np.pad(board, 1)

print(f"padded board  \n {p_board}")
print('\n')
print(f" p_board[0] \n {p_board[0]} ")
print('\n')
print(f"p_board[0][0] \n  {p_board[0][0]} ")
```

```python
print(f"padded board  \n {p_board}")
print(f"padded board dim \n {p_board.ndim}")
print(f"padded board shape \n {p_board.shape}")
print('\n')

r=1
c=1
print(p_board[r,c])
print("\n")

print( p_board[r-1:r+2, c-1:c+2] )
print("\n")
print(np.sum(p_board[r-1:r+2, c-1:c+2]))
```
