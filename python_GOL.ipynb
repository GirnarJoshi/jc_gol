```python
from jcgol import GameOfLife
```

```python
block = [
    (0,0), (1,0),
    (0,1), (1,1)
    ]

blinker = [
    (0,1) , (1,1), (2,1) 
]


glider = [
    (0, 2),
    (1, 0),
    (1, 2),
    (2, 1),
    (2, 2),
]
```

```python
gol = GameOfLife(25, 25, glider)
gol.run(20)
```

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

fig, ax = plt.subplots()

image = ax.imshow(
    gol.history_[0],
    interpolation="nearest"
)

ax.set_xticks([])
ax.set_yticks([])


def update(frame):
    image.set_data(gol.history_[frame])
    return [image]


animation = FuncAnimation(
    fig,
    update,
    frames=len(gol.history_),
    interval=100,
    blit=True
)

HTML(animation.to_jshtml())
```
